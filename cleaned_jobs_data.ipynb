```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Load original data
df = pd.read_csv("BMW.csv")

# Create a Streamlit app
st.title("BMW Sales Analysis Dashboard")

# Function to create and display charts
def create_charts(region):
    if region == 'North America':
        electric_chart = px.bar(df, x='region', y='electric_sales', title='Electric Sales by Region')
        petrol_chart = px.bar(df, x='region', y='petrol_sales', title='Petrol Sales by Region')
        diesel_chart = px.bar(df, x='region', y='diesel_sales', title='Diesel Sales by Region')
    elif region == 'Europe':
        electric_chart = px.bar(df, x='region', y='electric_sales', title='Electric Sales by Region')
        petrol_chart = px.bar(df, x='region', y='petrol_sales', title='Petrol Sales by Region')
        diesel_chart = px.bar(df, x='region', y='diesel_sales', title='Diesel Sales by Region')
    elif region == 'Asia':
        electric_chart = px.bar(df, x='region', y='electric_sales', title='Electric Sales by Region')
        petrol_chart = px.bar(df, x='region', y='petrol_sales', title='Petrol Sales by Region')
        diesel_chart = px.bar(df, x='region', y='diesel_sales', title='Diesel Sales by Region')

    # Create Plotly charts
    electric_chart_plotly = px.bar(df, x='region', y='electric_sales', title='Electric Sales by Region')
    petrol_chart_plotly = px.bar(df, x='region', y='petrol_sales', title='Petrol Sales by Region')
    diesel_chart_plotly = px.bar(df, x='region', y='diesel_sales', title='Diesel Sales by Region')

    # Display Matplotlib charts
    plt.figure(figsize=(10, 6))
    sns.set_style("whitegrid")
    sns.set_axis_labels("Region", "Sales")
    sns.set_legend_format("")
    plt.title(f"Electric Sales by Region (Plotly)")
    plt.xlabel("Region")
    plt.ylabel("Sales")
    plt.show()

    # Display Plotly charts
    st.plotly_chart(electric_chart_plotly)
    st.plotly_chart(petrol_chart_plotly)
    st.plotly_chart(diesel_chart_plotly)

# Function to update region selection
def update_region(region):
    create_charts(region)

# Main function to run the app
def main():
    region = st.selectbox("Select Region", df['Region'].unique())
    update_region(region)
    st.write(f"Region: {region}")

if __name__ == "__main__":
    main()
```
