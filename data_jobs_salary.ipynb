```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data jobs_salary.csv")
df.head()
```

```python
year_jobs = pd.pivot_table(data=df, index="work_year", columns="job_title", aggfunc="size")
year_jobs
```

```python
import seaborn as sns

sns.lineplot(data=year_jobs)
plt.xlabel("Work year")
plt.ylabel("Number of jobs")
plt.xticks(year_jobs.index)
plt.show()
```
