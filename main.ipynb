```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("job_data_cleaned.csv"); df.head()
```

```python
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df.iloc[0:5]
```

```python
df.loc[2:10, "Entry_Date":"job_posted_country"]
```

```python
median_salary_year = df["salary_year_average"].median()

median_salary_hour = df["salary_hour_average"].median()
```

```python
df["salary_year_average"] = df["salary_year_average"].fillna(median_salary_year)
df["salary_hour_average"] = df["salary_hour_average"].fillna(median_salary_hour)
```

```python
df_unique = df

df_unique = df_unique.drop_duplicates()
```

```python
print("Length of original df:", len(df))
print("Length of droped df:", len(df_unique))
print("Rows droped", len(df) - len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=["job_title", "job_posted_country"])

print("Length of original df:", len(df))
print("Length of droped df:", len(df_unique))
print("Rows droped", len(df) - len(df_unique))
```
